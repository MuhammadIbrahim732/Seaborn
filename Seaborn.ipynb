## **Seaborn Distribution Plots**

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.histplot(df["total_bill"], bins=20, kde = True)
plt.show()
```

```python
plt.subplot(1,2,1)
sns.histplot(df["total_bill"], bins=20, kde = True)
plt.subplot(1,2,2)
sns.histplot(df["tip"], bins=20, kde = True)
plt.show()
```

```python
marks = np.array([50,60,50,65,70,65,60,75,80,80,70,75,75,75,90])
sns.histplot(marks,bins=14,kde=True)
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

**Legends**

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.plot(x, y1, label="Sales")
plt.plot(x, y2, label="Profit")

plt.legend()
plt.show()
```

**Joint Plot**

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="hex")
plt.show()
```

**Scatter Plot**

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 6]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

**Pair Plot**

```python
sns.pairplot(df, hue = "sex")
plt.show()
```

```python
sns.pairplot(df, hue = "size", palette="rainbow")
plt.show()
```

**Rug Plot**

```python
sns.rugplot(df["total_bill"])
plt.show()
```

# **Seaborn Categorical Plots**

**Count Plot**

```python
sns.countplot(x = df['sex'], hue = df['smoker'])
plt.show()
```

**Bar Plot**

```python
sns.barplot(x = df["sex"], y = df["total_bill"])
plt.show()
```

```python
sns.barplot(x = df["sex"], y = df["total_bill"], estimator= np.sum)
plt.show()
```

```python
sns.barplot(x = df["sex"], y = df["tip"], estimator= np.sum)
plt.show()
```

**Box Plot**

```python
sns.boxenplot(x= df["tip"], y= df["day"],hue = df["sex"] , palette="rainbow")
plt.show()
```

**Violin Plot**

```python
sns.violinplot(x = df["day"], y = df["total_bill"])
plt.show()
```

**Strip Plot**

```python
sns.stripplot(x = df["day"], y = df["total_bill"])
plt.show()
```

**Swarm Plot**

```python
sns.swarmplot(x = df["day"], y = df["total_bill"])
plt.show()
```

**Line Chart**

```python
import matplotlib.pyplot as plt

month = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 150, 180]

plt.plot(month, sales)
plt.show()
```

**Bar Chart**

```python
import matplotlib.pyplot as plt

dept = ["IT", "HR", "Sales"]
emp = [30, 20, 25]

plt.bar(dept, emp)
plt.show()
```

**Pie Chart**

```python
import matplotlib.pyplot as plt

labels = ["IT", "HR", "Sales"]
sizes = [30, 20, 50]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.show()
```

# **Seaborn Matrix Plots**

**Heatmap**

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset("flights")
flights
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
tips_corr = tips[["total_bill","tip","size"]]
tips_corr
```

```python
tips_corr.corr()
```

```python
sns.heatmap(tips_corr.corr(), annot = True)
plt.show()
```

**Cluster Map**

```python
sns.clustermap(tips_corr.corr())
plt.show()
```
